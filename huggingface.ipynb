```python
import torch
from transformers import AutoModel, pipeline
from transformers.pipelines import PIPELINE_REGISTRY

from vocal_remover.transformer.configuration import VocalRemoverConfig
from vocal_remover.transformer.modeling import VocalRemoverModel
from vocal_remover.transformer.pipeline import VocalRemoverPipeline
```

```python
vocal_remover_config = VocalRemoverConfig()
```

```python
vocal_remover_model = VocalRemoverModel(vocal_remover_config)
vocal_remover_model.model.load_state_dict(
    torch.load("src/vocal_remover/models/baseline.pth")
)
```

```python
VocalRemoverConfig.register_for_auto_class()
VocalRemoverModel.register_for_auto_class()
```

```python
vocal_remover_model.push_to_hub("tsurumeso-vocal-remover")  # pyright: ignore[reportArgumentType]
```

```python
PIPELINE_REGISTRY.register_pipeline(
    "tsurumeso-vocal-remover",
    pipeline_class=VocalRemoverPipeline,
    pt_model=AutoModel,
    default={"model": ("NextFire/tsurumeso-vocal-remover", "main")},
    type="audio",
)
```

```python
# type: ignore
vocal_remover_pipeline = pipeline(
    task="tsurumeso-vocal-remover",
    trust_remote_code=True,
)
```

```python
vocal_remover_pipeline.push_to_hub("tsurumeso-vocal-remover")
```
